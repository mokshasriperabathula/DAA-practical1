```python
from collections import deque

# Graph represented as an adjacency list
graph = {
    'A': ['B', 'C'],
    'B': ['D', 'E'],
    'C': ['F'],
    'D': []
}
print(graph)
```

```python
from collections import deque

def bfs_search(graph, start):
    visited = set()
    queue = deque([start])

    visited.add(start)

    while queue:
        vertex = queue.popleft()
        print(vertex, end=" ")

        for neighbor in graph.get(vertex, []):
            if neighbor not in visited:
                visited.add(neighbor)
                queue.append(neighbor)
```

```python
print("BFS Traversal starting from node 'A':")
bfs_search(graph, 'A')
```

```python
def dfs(graph, node,visited=None):
  if visited is None:
    visited = set()
  visited.add(node)
  print(node, end=" ")
  for neighbor in graph.get(node, []):
    if neighbor not in visited:
      dfs(graph, neighbor, visited)

print("DFS Traversal starting from node 'A':")
dfs(graph, 'A')
```

```python
graph = {
    'A': [('B', 2), ('C', 4)],
    'B': [('D', 3), ('E', 5)],
    'C': [('E', 1)],
    'D': [('F', 2)],
    'E': [('F', 3)],
    'F': []
}

def path_cost(path):
    cost = 0

    for i in range(len(path) - 1):
        current = path[i]
        next_node = path[i + 1]

        for neighbor, edge_cost in graph[current]:
            if neighbor == next_node:
                cost += edge_cost
                break

    return cost


path1 = ['A', 'C', 'E', 'F']
path2 = ['A', 'C', 'F']

print("Path:", path1)
print("Path:", path2)
print("Path Cost:", path_cost(path1))
print("Path Cost:", path_cost(path2))
```

```python
cost1 = path_cost(path1)
cost2 = path_cost(path2)

print(f"Cost of {path1}: {cost1}")
print(f"Cost of {path2}: {cost2}")

if cost1 < cost2:
    print(f"Path {path1} is cheaper with a cost of {cost1}.")
elif cost2 < cost1:
    print(f"Path {path2} is cheaper with a cost of {cost2}.")
else:
    print("Both paths have the same cost.")
```
